```python
import numpy as np

# 3.1 Create the Image Matrix
img = np.arange(36).reshape(6, 6)

print("Original Image:")
print(img)


# 3.2 Extract the Center Region
center = img[1:5, 1:5]

print("\nCenter Region:")
print(center)


# 3.3 Apply Brightness Enhancement
bright_center = center + 10

print("\nBrightened Center:")
print(bright_center)


# 3.4 Update the Original Image
img[1:5, 1:5] = bright_center

print("\nUpdated Image:")
print(img)


# 3.5 Compute Image Statistics
print("\nImage Statistics:")
print("Mean Intensity:", img.mean())
print("Max Intensity:", img.max())
print("Min Intensity:", img.min())


# Question 4
# Decrease brightness of border pixels by 5
img[0, :] -= 5
img[-1, :] -= 5
img[:, 0] -= 5
img[:, -1] -= 5

print("\nFinal Image after decreasing border brightness:")
print(img)


# Question 5
# Compute Standard Deviation
print("\nStandard Deviation:", img.std())
```
